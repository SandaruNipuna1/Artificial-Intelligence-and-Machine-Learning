```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, silhouette_score
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/heart_cleaned_no_outliers_encoded_Featured_Standardized(1).csv')
```

```python
X = df.drop('target', axis=1)  # Features
y = df['target']  # Target variable
```

```python
print(f"Dataset shape: {df.shape}")
print(f"Features: {X.shape[1]}, Target: {y.shape[0]}")
```

```python

kmeans1 = KMeans(n_clusters=2, init='k-means++', max_iter=300, n_init=10, random_state=42)
clusters1 = kmeans1.fit_predict(X)
```

```python
# Evaluate clustering
def cluster_to_labels(clusters, true_labels):
    cluster_labels = {}
    for cluster in np.unique(clusters):
        mask = clusters == cluster
        if np.sum(mask) > 0:
            cluster_labels[cluster] = np.bincount(true_labels[mask]).argmax()
    predicted_labels = np.array([cluster_labels[c] for c in clusters])
    return predicted_labels

predicted_labels1 = cluster_to_labels(clusters1, y)
```

```python

# Calculate metrics
accuracy1 = accuracy_score(y, predicted_labels1)
precision1 = precision_score(y, predicted_labels1)
recall1 = recall_score(y, predicted_labels1)
f1_1 = f1_score(y, predicted_labels1)
silhouette1 = silhouette_score(X, clusters1)

print("\n Variety 1 Results:")
print(f"Accuracy: {accuracy1:.4f} ({accuracy1*100:.2f}%)")
print(f"Precision: {precision1:.4f}")
print(f"Recall: {recall1:.4f}")
print(f"F1-Score: {f1_1:.4f}")
print(f"Silhouette Score: {silhouette1:.4f}")
```

```python
#variety 2
df_pca = pd.read_csv('/content/heart_dataset_pca.csv')
```

```python
print(f"PCA dataset shape: {df_pca.shape}")
print("PCA features:", df_pca.columns.tolist())
```

```python
X_pca = df_pca.drop('target', axis=1)
y_pca = df_pca['target']


print(f"PCA features shape: {X_pca.shape}")
print(f"Target distribution:\n{y_pca.value_counts()}")
```

```python
# apply k-means on PCA data
kmeans2 = KMeans(n_clusters=2, init='random', max_iter=300, n_init=10, random_state=42)
clusters2 = kmeans2.fit_predict(X_pca)
```

```python
# convert clusters to labels
predicted_labels2 = cluster_to_labels(clusters2, y_pca)
```

```python
# Calculate metrics
accuracy2 = accuracy_score(y_pca, predicted_labels2)
precision2 = precision_score(y_pca, predicted_labels2)
recall2 = recall_score(y_pca, predicted_labels2)
f1_2 = f1_score(y_pca, predicted_labels2)
silhouette2 = silhouette_score(X_pca, clusters2)

print("\n Variety 2 Results:")
print(f"Accuracy: {accuracy2:.4f} ({accuracy2*100:.2f}%)")
print(f"Precision: {precision2:.4f}")
print(f"Recall: {recall2:.4f}")
print(f"F1-Score: {f1_2:.4f}")
print(f"Silhouette Score: {silhouette2:.4f}")
```

```python
print(" Variety 3: Feature-Selected Data")
selected_features = ['oldpeak', 'exang_encoded', 'slope_encoded', 'thal_encoded', 'thalach', 'cp_encoded']
```

```python
available_features = [f for f in selected_features if f in X.columns]
```

```python
if len(available_features) > 0:
    X_selected = X[available_features]
    print(f"Selected features: {available_features}")
```

```python
# Apply K-Means on feature-selected data
kmeans3 = KMeans(n_clusters=2, init='k-means++', max_iter=300, n_init=10, random_state=42)
clusters3 = kmeans3.fit_predict(X_selected)
```

```python
# Convert clusters to labels
predicted_labels3 = cluster_to_labels(clusters3, y)
```

```python
# Calculate metrics
accuracy3 = accuracy_score(y, predicted_labels3)
precision3 = precision_score(y, predicted_labels3)
recall3 = recall_score(y, predicted_labels3)
f1_3 = f1_score(y, predicted_labels3)
silhouette3 = silhouette_score(X_selected, clusters3)

print("\n Variety 3 ")
print(f"Accuracy: {accuracy3:.4f} ({accuracy3*100:.2f}%)")
print(f"Precision: {precision3:.4f}")
print(f"Recall: {recall3:.4f}")
print(f"F1-Score: {f1_3:.4f}")
print(f"Silhouette Score: {silhouette3:.4f}")
```

```python
# Summary of all varieties
print(" K-MEANS RESULTS SUMMARY")


results = {
    'Variety 2 (PCA)': [accuracy2, precision2, recall2, f1_2, silhouette2],
    'Variety 1 (Standard)': [accuracy1, precision1, recall1, f1_1, silhouette1],
    'Variety 3 (Feature-Selected)': [accuracy3, precision3, recall3, f1_3, silhouette3]
}

summary_df = pd.DataFrame(results,
                         index=['Accuracy', 'Precision', 'Recall', 'F1-Score', 'Silhouette'])
print(summary_df.round(4))
```

```python
# Find best performing variety
best_variety = max(results.keys(), key=lambda x: results[x][0])  # Based on accuracy
print(f"\n Best Performing Variety: {best_variety}")
print(f"Best Accuracy: {results[best_variety][0]:.4f} ({results[best_variety][0]*100:.2f}%)")
```

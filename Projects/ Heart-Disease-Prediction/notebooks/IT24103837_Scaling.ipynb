```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import numpy as np

# Load the dataset
df = pd.read_csv('heart_cleaned_no_outliers_encoded_Featured.csv')

# Separate features and target variable
X = df.drop('target', axis=1)  # Features
y = df['target']  # Target variable

# Initialize the StandardScaler
scaler = StandardScaler()

# Standardize all features
X_standardized = scaler.fit_transform(X)

# Convert back to DataFrame with original column names
X_standardized_df = pd.DataFrame(X_standardized, columns=X.columns)

# Add the target column back
df_standardized = pd.concat([X_standardized_df, y.reset_index(drop=True)], axis=1)

# Display the first few rows to verify standardization
print("Original dataset shape:", df.shape)
print("Standardized dataset shape:", df_standardized.shape)
print("\nFirst 5 rows of standardized dataset:")
print(df_standardized.head())

# Display statistics to confirm standardization
print("\nStatistics of standardized features:")
print(df_standardized.describe())

# Save the standardized dataset to a new CSV file
df_standardized.to_csv('heart_cleaned_no_outliers_encoded_Featured_Standardized.csv', index=False)
print("\nStandardized dataset saved to 'heart_cleaned_no_outliers_encoded_Featured_Standardized.csv'")
```

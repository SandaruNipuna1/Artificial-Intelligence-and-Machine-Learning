```python
import pandas as pd

df = pd.read_csv('heart.csv')

missing_values = df.isnull().sum()

columns_with_missing = missing_values[missing_values > 0]

print("Columns with missing values:")
print("----------------------------")
for column, count in columns_with_missing.items():
    print(f"{column}: {count} missing values")

print("\nAll columns with missing value counts:")
print("-------------------------------------")
print(missing_values)

print("\nPercentage of missing values:")
print("-----------------------------")
for column, count in missing_values.items():
    if count > 0:
        percentage = (count / len(df)) * 100
        print(f"{column}: {count} ({percentage:.2f}%)")
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
missing_values.sort_values(ascending=False).plot.bar(color="red")
plt.title("Missing Values per Column")
plt.ylabel("Number of Missing Values")
plt.xlabel("Columns")
plt.show()
```

```python
import pandas as pd

df = pd.read_csv("heart.csv")

for col in ["age", "trestbps", "chol", "thalach"]:
    df[col] = df[col].fillna(round(df[col].mean())).astype(int)

df["sex"] = df["sex"].fillna(df["sex"].mode()[0])

# Check if missing values remain
print("\nMissing values after imputation:")
print(df.isnull().sum())
```

```python
df.to_csv("heart_cleaned.csv", index=False)
```
